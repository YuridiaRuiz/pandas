#### Pandas DataTypes
- series (`s`) - single dimensional array. Single column
    - homogenous - 1 data type per series
- dataframe (`df`) - two dimentional array (table). collection of colums

```python
# declare a series
ser = pd.Series(list('abcdfghijklmnopqrstuvwxyz'))

ser.name = 'alphabet'

# .head(n) - show top 5 records - n =number f records
print(ser.head())

# tail(n) - shop the bottom 5 records - n = number of records
print(ser.tail())
```

```python
# Declare a series from a dictionary

d={
    'a':0.0,
    'b':1.0,
    'c':2.0,
}

# For a dictionary becoming a series, the keys become the index 
float_series = pd.Series(d)

print(float_series)
```

```python
# Index is specified, corresponding data will be pulled into the series
customer_index = pd.Series(d, index=['a', 'y', 'x'])

print(customer_index)
```

```python
carlories = {
    'day1': 2200,
    'day2': 2500,
    'day3': 2700,
}


#Create a series with the following (or similare dictionay), but only with days 1 and 2
index=['day1', 'day2']
calories_series = pd.Series(carlories, index=index)
print(calories_series)
```

```python
#Series of scalar value, Each index will have the same value
scalar_series = pd.Series(5.0, index=['a','b','c','d','e'])
print(scalar_series)
```

```python
ser1 = pd.Series([2, 4, 6, 8, 10])
ser2 = pd.Series([1, 3, 5, 7, 9])
#Just like arrays, math with series are element-wise
print(ser1 + ser2)
print(ser1 - ser2)
print(ser1 * ser2)
print(ser1 / ser2)
```

```python
#Compare series
#Element - wise arithmetic with series
ser1 = pd.Series([2, 4, 6, 8, 10])
ser2 = pd.Series([1, 3, 5, 7, 9])

print(ser1 > ser2)
```
