#### Pandas DataTypes
- series (`s`) - single dimensional array. Single column
    - homogenous - 1 data type per series
- dataframe (`df`) - two dimentional array (table). collection of colums

```python
# declare a series
ser = pd.Series(list('abcdfghijklmnopqrstuvwxyz'))

ser.name = 'alphabet'

# .head(n) - show top 5 records - n =number f records
print(ser.head())

# tail(n) - shop the bottom 5 records - n = number of records
print(ser.tail())
```

```python
# Declare a series from a dictionary

d={
    'a':0.0,
    'b':1.0,
    'c':2.0,
}

# For a dictionary becoming a series, the keys become the index 
float_series = pd.Series(d)

print(float_series)
```

```python
# Index is specified, corresponding data will be pulled into the series
customer_index = pd.Series(d, index=['a', 'y', 'x'])

print(customer_index)
```

```python
carlories = {
    'day1': 2200,
    'day2': 2500,
    'day3': 2700,
}


#Create a series with the following (or similare dictionay), but only with days 1 and 2
index=['day1', 'day2']
calories_series = pd.Series(carlories, index=index)
print(calories_series)
day_index_info = calories_series.index.tolist()

print(day_index_info)
```

```python
#Series of scalar value, Each index will have the same value
scalar_series = pd.Series(5.0, index=['a','b','c','d','e'])
print(scalar_series)
```

```python
ser1 = pd.Series([2, 4, 6, 8, 10])
ser2 = pd.Series([1, 3, 5, 7, 9])
#Just like arrays, math with series are element-wise
print(ser1 + ser2)
print(ser1 - ser2)
print(ser1 * ser2)
print(ser1 / ser2)
```

```python
#Compare series
#Element - wise arithmetic with series
ser1 = pd.Series([2, 4, 6, 8, 10])
ser2 = pd.Series([1, 3, 5, 7, 9])

print(ser1 > ser2)
```

### SLicing a Data Frame
- pull out specific parts of a dataframe
- examine a subset
- removeunwanted/unneeded data
- data isolation, remove clutter
- `df[[col1,col2,col3,col4]]` - slicing using double square bracket syntax.


```python
# Initializing the nested list with Data set
employee_list = [['James', 36, 75, 5428000],
               ['Villers', 38, 74, 3428000],
               ['VKole', 31, 70, 8428000],
               ['Smith', 34, 80, 4428000],
               ['Gayle', 40, 100, 4528000],
               ['Rooter', 33, 72, 7028000],
               ['Peterson', 42, 85, 2528000],
               ['John', 41, 85, 1528000],

]

columns = ['Name','Age','Weight (Kgs)','Salary']

df = pd.DataFrame(employee_list,columns=columns)
print(df.head())
```

### Slicing Function
- `df.loc[start:stop:step(rows), col]` - label based location() -label based location (column names to locate and slice)
- `df.iloc[star:stop:step (rows), start:stop:step (cols)]` -  index based location (index numbers to locate and slice)
- with th python indexing operator. STOP IS ALWAYS NON_INCLUSIVE

```python
# python indexing operator
emp_subset = df.iloc[:,0:2]
print(emp_subset)
```

```python
# Using iloc, how can you grab the first 2 records of the first 2 columns

emp_sub_subset = df.iloc[:2,:2]
print(emp_sub_subset)
```

```python
# use a list in the column argument to specify specific list
print(df.iloc[:,[0,2,3]])
```

### Dealing with NaN
- `np.nan`
- `dropna`

```python
# Initializing the nested list with Data set
employee_list = [['James', 36, 75, 5428000],
               ['Villers', 38, 74, 3428000],
               ['VKole', 31, 70, 8428000],
               ['Smith', 34, 80, 4428000],
               ['Gayle', 40, 100, 4528000],
               ['Adam', 40, np.nan, 4528000],
               ['Rooter', 33, 72, 7028000],
               ['Peterson', 42, 85, 2528000],
               ['lynda', 42, 85, np.nan],
               [np.nan, 42, 85, np.nan],
               ['Jenny', np.nan, 100, 25632],
               ['Kenn', np.nan, 110, 25632],
                ['Aly', np.nan, 90, 25582],
               ['John', 41, 85, 1528000]]

columns=['Name', 'Age', 'Weight', 'Salary']

df = pd.DataFrame(employee_list, columns=columns)
```

```python
print(f'Missing values: \n{df.isnull().sum()}')
```

```python
df = df.dropna()
print(f'Missing values: \n{df.isnull().sum()}')
```
